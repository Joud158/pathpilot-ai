### Live demonstration: leads, unknown questions, feedback, and memory
Run with a configured API key. This uses fictional contact details and creates actual tool/reflection logs only if the model calls the tools. Inspect the files to verify behavior; these are live calls, not the offline scripted tests.


```python
if os.getenv("GEMINI_API_KEY" if PROVIDER == "gemini" else "OPENAI_API_KEY"):
    history = []
    for question in [
        "I'm interested in a mock interview. Please have your team follow up.",
        "My name is Alex Lee and my email is alex@example.com. Please contact me about that service.",
        "Do you offer refunds?",
        "I have feedback: please add weekend appointments.",
    ]:
        answer = respond(question, history)
        print("USER:", question, "\nNOVA:", answer, "\n")
        history.extend([{"role": "user", "content": question}, {"role": "assistant", "content": answer}])
    for filename in ["leads.jsonl", "feedback.jsonl", "reflection.jsonl"]:
        log = Path("logs") / filename
        print(filename, log.read_text(encoding="utf-8") if log.exists() else "No records yet; check the model's tool calls.")
else:
    print("Configure your API key before running the live demonstration.")
```

## 9. Automated tests
Uses the notebook kernel Python and a unique project-local temporary directory to avoid Windows temporary-folder permission issues.


```python
import subprocess, sys, uuid
from pathlib import Path
test_temp = Path(".test_tmp") / uuid.uuid4().hex
test_temp.parent.mkdir(parents=True, exist_ok=True)
result = subprocess.run([sys.executable, "-m", "pytest", "-q", "tests", "--basetemp", str(test_temp), "--color=no"], capture_output=True, text=True)
print(result.stdout)
print(result.stderr)
assert result.returncode == 0, "Tests failed; inspect the output above."
```
